# Pupillary Light Response Simulation

This example simulates a measurement with `piprkit.simulate_pupil_measurement`. It starts with a default simulation, and then slowly expands with more options.

```python
import matplotlib.pyplot as plt

import piprkit
```

## 1. Basic Pupil Response Simulation

`simulate_pupil_measurement` uses the model as explained in [fitting examples](../../pupil_fit.rst). There are four phases:
- Baseline
- Constriction
- Maintained constriction
- Redilation

Default values are illustrative only; they might not result in a realistic example.

```python
pupil_measurement = piprkit.simulate_pupil_measurement()

fig, ax = plt.subplots(figsize=(8, 2))

pupil_measurement.plot(
    ax=ax,
    color="blue",
    linewidth=2,
    label="Pupil Size"
)

pupil_measurement.plot_light_stimulus(
    ax=ax,
    color="yellow",
    alpha=0.3,
    label="Light Stimulus"
)

ax.set_title("Simulated Pupil Response")
ax.legend()
plt.show()
```

## 2. Customizing Simulation Parameters

Various time-related aspects of the simulation can be controlled.

### Timing Parameters:
- `t_start`, `t_end`: Overall simulation time window
- `t_light_onset`, `t_light_offset`: Light stimulus timing
- `num_samples`: Number of data samples

```python
pupil_measurement = piprkit.simulate_pupil_measurement(
    t_start=0.0,
    t_end=16.0,
    t_light_onset=2.0,
    t_light_offset=6.0,
    num_samples=1000
)

fig, ax = plt.subplots(figsize=(8, 2))

pupil_measurement.plot(
    ax=ax,
    color="blue",
    linewidth=2,
    label="Pupil Size"
)

pupil_measurement.plot_light_stimulus(
    ax=ax,
    color="yellow",
    alpha=0.3,
    label="Light Stimulus"
)

ax.set_title("Simulated Pupil Response with Custom Timing")
ax.legend()
plt.show()
```

## 3. Customizing Response Parameters

The parameters used to simulate the data can be customized:
- `constriction_delay`: the time from light onset to the constriction start, in seconds.
- `time_to_peak`: the time from light onset to the peak constriction, in seconds.
- `redilation_delay`: the time from light offset to the redilation start, in seconds.
- `baseline_c`: the baseline size.
- `maintained_constriction_c`: the size at the peak constriction.
- `maintained_constriction_m`: the slope of the maintained constriction, in size units per second.
- `redilation_k`: the rate constant k of the redilation exponential, in 1/s.
- `redilation_P`: the plateau P of the redilation exponential: the size it approaches.

```python
pupil_measurement = piprkit.simulate_pupil_measurement(
    baseline_c=7.5,
    maintained_constriction_c=3.9,
    maintained_constriction_m=0.1,
    redilation_k=-0.5,
    redilation_P=6.2,
)

fig, ax = plt.subplots(figsize=(8, 2))

pupil_measurement.plot(
    ax=ax,
    color="blue",
    linewidth=2,
    label="Pupil Size"
)

pupil_measurement.plot_light_stimulus(
    ax=ax,
    color="yellow",
    alpha=0.3,
    label="Light Stimulus"
)

ax.set_title("Simulated Pupil Response with Custom Parameters")
ax.set_ylim(2, 8)
ax.legend()
plt.show()
```

## 4. Comparing Different Stimulus Conditions

```python
red_stim = piprkit.simulate_pupil_measurement(
    maintained_constriction_c=0.64,
)

blue_stim = piprkit.simulate_pupil_measurement(
    maintained_constriction_c=0.54,
)

red_peak = red_stim.metrics.peak_constriction()
blue_peak = blue_stim.metrics.peak_constriction()

print("Stimulus comparison:")
print(f"Red stimulus - {red_peak.get_name()}: {red_peak:.2f}")
print(f"Blue stimulus - {blue_peak.get_name()}: {blue_peak:.2f}")

fig, ax = plt.subplots(1, 1, figsize=(8, 2))

red_stim.plot(
    ax=ax,
    color="red",
    linewidth=2,
    label="Red stimulus"
)

red_stim.plot_light_stimulus(
    ax=ax,
    color="yellow",
    alpha=0.3
)

blue_stim.plot(
    ax=ax,
    color="blue",
    linewidth=2,
    label="Blue stimulus"
)

red_peak.plot(ax=ax, color="darkred", label=f"Red {red_peak.get_label()}")
blue_peak.plot(ax=ax, color="darkblue", label=f"Blue {blue_peak.get_label()}")

ax.set_title("Light Stimulus comparison")
ax.legend()
plt.show()
```

## 5. Adding Noise

Noise can be added to make the data appear more realistic. `noise_std` sets the amount of Gaussian noise, `trend_std` sets the amount of cumulative noise. An rng seed can be set.

```python
clean_signal = piprkit.simulate_pupil_measurement(
    noise_std=0.0,
    trend_std=0.0
)

noisy_signal = piprkit.simulate_pupil_measurement(
    noise_std=0.001,
    trend_std=0.001,
    rng=0
)
very_noisy_signal = piprkit.simulate_pupil_measurement(
    noise_std=0.005,
    trend_std=0.005,
    rng=0
)

fig, ax = plt.subplots(1, 1, figsize=(8, 2))

clean_signal.plot(
    ax=ax,
    color="green",
    linewidth=1.5,
    label="No noise"
)

noisy_signal.plot(
    ax=ax,
    color="orange",
    linewidth=1.5,
    label="Moderate noise"
)
very_noisy_signal.plot(ax=ax, color="red", linewidth=1.5, label="High noise")

clean_signal.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3)

ax.legend()
plt.show()
```
