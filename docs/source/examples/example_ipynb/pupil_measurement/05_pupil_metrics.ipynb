# Pupil Metrics

There are many metrics that can be used. This example page shows a few. Detailed info can be found in [API reference](../../../api.rst), as well as more metrics.

```python
from matplotlib import pyplot as plt
import piprkit
```

## Baseline

`metrics.baseline(duration)` returns the average pupil size throughout the baseline period.

```python
pm = piprkit.load_simulated_pupil()
baseline = pm.metrics.baseline(duration=5)
print(baseline)

fig, ax = plt.subplots(1, 1, figsize=(8, 2))
pm.plot(ax=ax, label="Pupil Size", color="blue")
pm.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3)
baseline.plot(ax=ax, color="red", label=f"{baseline.get_label()}: {baseline:.2f}")
ax.legend()
plt.show()
```

## 6s-PIPR

The PIPR metrics such as 6s-PIPR can be retrieved too.

```python
pm = piprkit.load_simulated_pupil()
# The average size can be determined from the pupil size data.
average_size_6s = pm.metrics.get_average_size(start_time=5.95, end_time=6.05)
print(f"Average pupil size at 6 seconds post-stimulus: {average_size_6s:.2f} mm")
# Getting the 6s-pipr is very similar to getting the average size, but uses predefined time windows.
pipr_6s = pm.metrics.pipr_6s()
print(f"{pipr_6s.get_name()}: {pipr_6s:.2f}")

fig, ax = plt.subplots(1, 1, figsize=(8, 2))
pm.plot(ax=ax, label="Pupil Size", color="blue")
pm.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3)
pipr_6s.plot(ax=ax, color="green", label=f"{pipr_6s.get_label()}: {pipr_6s:.2f}")
ax.legend()
plt.show()
```

## Metrics on baseline-corrected data

If a metric is determined after baseline correction, the metric will reflect this correction.

```python
pm = piprkit.load_simulated_pupil()
pm.filters.apply_baseline_correction(duration=5)
print(pm.metrics.peak_constriction())
print(pm.metrics.pipr_6s())
```

## PLR latency

`plr_latency` returns the time to reach 1% constriction.

```python
sim = piprkit.simulate_pupil_measurement(t_start=-12, t_end=40, num_samples=5201)
sim.filters.apply_baseline_correction(duration=10)
plr_latency = sim.metrics.plr_latency()
print(plr_latency)

fig, ax = plt.subplots(1, 1, figsize=(8, 2))
sim.plot(ax=ax, color="blue", label="Pupil size")
sim.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3)
plr_latency.plot(ax=ax, color="green", label=f"{plr_latency.get_label()}: {plr_latency:.2f} s")
ax.set_xlim(-2, 2)
ax.legend()
plt.show()
```

## AUC and PIPR duration

The AUC can be determined, as well as the pipr duration.

```python
auc_early = sim.metrics.auc_early()
auc_late = sim.metrics.auc_late()
pipr_duration = sim.metrics.pipr_duration()
print(auc_early)
print(auc_late)
print(pipr_duration)

fig, ax = plt.subplots(1, 1, figsize=(8, 2))
sim.plot(ax=ax, color="blue", label="Pupil size")
sim.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3)
auc_early.plot(ax=ax, color="red", label=f"{auc_early.get_label()}: {auc_early:.2f}")
auc_late.plot(ax=ax, color="purple", label=f"{auc_late.get_label()}: {auc_late:.2f}")
pipr_duration.plot(ax=ax, color="green", label=f"{pipr_duration.get_label()}: {pipr_duration:.1f} s")
ax.legend()
plt.show()
```

## Net PIPR

`piprkit.net_pipr(a, b)` returns the net pipr between metric `a` and `b`.

```python
blue = piprkit.load_simulated_pupil("blue")
red = piprkit.load_simulated_pupil("red")
blue.filters.apply_baseline_correction(duration=5)
red.filters.apply_baseline_correction(duration=5)
net = piprkit.net_pipr(blue.metrics.pipr_6s(), red.metrics.pipr_6s())
print(net)
```

## Metrics from a fitted model

Some metrics can only be determined based on a model fit.

```python
model = piprkit.PhaseModel(sim)
print(model.metrics.constriction_v())
print(model.metrics.pupil_escape())
print(model.metrics.redilation_v())
print(model.metrics.plateau_pipr())
```
