# Series loading

Whereas `PupilMeasurement` contained data on a single measurement, we often have many measurements in a row. We use `PupilSeries` for this. It has large overlap with `PupilMeasurement` in available functions; it can use most of the filters. Using `split()`, the `PupilSeries` can be split into individual `PupilMeasurement`.

```python
import piprkit
import matplotlib.pyplot as plt
import numpy as np
```

## The example dataset

We load an example data set, containing real-world measured data.

```python
ps = piprkit.load_real_series()

time = ps.get_time()
print(f"Series data: {len(time)} time points")
print(f"Time range: {time[0]:.1f}s to {time[-1]:.1f}s")
print(f"Median sampling interval: {np.median(np.diff(time)) * 1000:.1f} ms")

stimuli_times = ps.get_light_stimuli_series().get_timings()
print(f"Number of light stimuli: {len(stimuli_times)}")
for i, (start, end) in enumerate(stimuli_times):
    print(f"Stimulus {i + 1}: {start:.1f}s - {end:.1f}s (duration: {end - start:.2f}s)")
```

## Plotting the series

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 2))

ps.plot(ax=ax, label="Pupil Trace", color="darkblue", alpha=0.8, linewidth=1)
ps.plot_light_stimulus(ax=ax, color="yellow", alpha=0.3, label="_Light Stimuli")

ax.set_title("Complete Pupillometry Series with Multiple Light Stimuli")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Pupil Size (mm)")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

pupil_data = ps.get_size()
print(f"\nSeries Statistics:")
print(f"Mean pupil size: {np.nanmean(pupil_data):.3f} mm")
print(f"Standard deviation: {np.nanstd(pupil_data):.3f} mm")
print(f"Range: {np.nanmin(pupil_data):.3f} - {np.nanmax(pupil_data):.3f} mm")
print(f"Recording duration: {ps.get_time()[-1] - ps.get_time()[0]:.1f} seconds")
```

## Building a series from arrays

Similar to `PupilMeasurement`, the `PupilSeries` can be made from array-like objects. It does require a `stimuli_list` with `(start, end)` to be defined.

```python
time_data = ps.get_time()
size_data = ps.get_size()
stimuli_list = [
    (8.06, 9.23),
    (68.41, 69.58),
    (128.73, 129.90),
    (189.08, 190.25),
    (249.40, 250.57),
    (309.75, 310.92),
]

own_series = piprkit.PupilSeries(time_data=time_data, size_data=size_data, stimuli_list=stimuli_list)
print(f"{len(own_series.get_time())} time points")
print(f"Light stimuli: {own_series.get_light_stimuli_series().get_timings()}")
```
